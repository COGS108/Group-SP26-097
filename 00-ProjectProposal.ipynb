## Research Question

Music is widely recognized as having meaningful effects on human emotion, cognition, and overall wellbeing. In this project, we aim to investigate **how music affects individuals with mental disorders, how it influences subjective productivity, and whether generalizable patterns emerge that describe music’s broader impact on the human mind**.

Our dataset will be constructed from extensive survey responses, including features such as preferred genres, listening habits, subjective productivity levels, and mental‑health classifications. Using these variables, we will conduct both exploratory and inferential analyses to identify patterns, correlations, and potential causal relationships. Ultimately, we hope to determine whether specific musical characteristics or genre preferences can meaningfully inform our understanding of mental health and productivity at a population level.
