Notice that the number of non-null in `thought response`, `arousal`, and `mood` features are the same within both experiments. This can mean that there are valuable insights we can develop from that sample population within the study.

However, we will first attempt to combine the two experiments together by first analyzing their distributions to see if they are similar. If they are not similar, we will have to analyze these two dataset separately and treat them as different sampling populations.

Because there are some considerations with the proportions of `thought response/probe`, `arousal`, and `mood` in each experiment, we will temporarily have 1 full dataset of complete responses that are only from participants who were randomly selected to answer these 3 questions.

```python
# Add experiment indicator to each before concatenating
exp_1_df['experiment'] = 1
exp_2_df['experiment'] = 2

# Rename columns to match (exp_2 uses slightly different names)
exp_2_df = exp_2_df.rename(columns={
    'reaction time': 'Reaction time',
    'no_go': 'No-go trial',
    'thought_response': 'Thought probe ',
    'arousal': 'Arousal',
    'mood': 'Mood',
    'music-present': 'Music-present'
})

# Concatenate
combined_df = pd.concat([exp_1_df, exp_2_df], ignore_index=True)
combined_df.info()
```
