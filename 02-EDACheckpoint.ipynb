# Research Question

Music is widely recognized as having meaningful effects on human emotion, cognition, and overall wellbeing. In this project, we aim to investigate **how music affects individuals with mental disorders, how it influences subjective productivity, and whether generalizable patterns emerge that describe music’s broader impact on the human mind**.

Our dataset will be constructed from extensive survey responses, including features such as preferred genres, listening habits, subjective productivity levels, and mental‑health classifications. Using these variables, we will conduct both exploratory and inferential analyses to identify patterns, correlations, and potential causal relationships. Ultimately, we hope to determine whether specific musical characteristics or genre preferences can meaningfully inform our understanding of mental health and productivity at a population level.

## Background and Prior Work

The effect of music on the brain has long been studied. For generations, music has been enjoyed and created by diverse cultures. It connects people, it amplifies emotions, and it varies in style to appeal to a wide range of people[<sup>1</sup>](#fn1). Some also claim to prefer working alongside music rather than without it. However, music’s effect on productivity may depend on how it is used.  This analysis, therefore, focuses on which genre, if any, is most conducive to productivity. 

Positive correlations between mental health and music have been recorded across multiple studies. There is a general consensus that enjoyment of music can reduce mental stress, anxiety, and depression among other difficult conditions. For example, Australian researchers Melissa K. Weinberg and Dawn Joseph found that in their sample, subjective wellbeing for individuals who had higher engagement with music was higher than those without[<sup>3</sup>](#fn3). This finding reveals a positive relationship between music engagement and emotional wellbeing and suggests our findings should coincide with this trend. However, this leaves a gap in specificity, as the study does not distinguish between types of music. 

Prior work hints that cognitive performance may be enhanced by slow-tempo music[<sup>2</sup>](#fn2). In this study, participants were driving and performing attention-demanding tasks. Findings showed that when listening to slow music, they responded faster to stimuli than with brown noise, silence, or other energetic music[<sup>2</sup>](#fn2). These results suggest that soothing genres like jazz and lo-fi may be associated with improved performance compared to louder, more energetic ones such as metal and rock. The article, however, does introduce an important nuance where its findings conflict with other such research and music was found to instead impede performance[<sup>2</sup>](#fn2). Thus, the question of whether music affects performance may depend on the specific task being measured rather than productivity in general. Therefore, “productivity” must be more precisely defined when conducting data analysis.
    
If we focus on the population of young adults aged 18-25, who regularly use music during daily activities, we can observe a similar phenomenon. In a study conducted on aproximately 100 college students, researchers found that music with lyrics impaired cognitive abilities[<sup>4</sup>](#fn4). This aligns with previous studies where music with more stimuli reduced focus. However, while students performed better on tasks with instrumental music than music with lyrics, they did not perform significantly better or worse than when in silence[<sup>4</sup>](#fn4). This suggests that while young adults should fare better with certain genres, these genres should not be recommended for enhanced performance overall. Rather, they should be recommended only in relation to other genres.
    
Ultimately, prior research on the influence of music on the human mind suggests a largely positive impact. Music may provide relaxation, emotional expression, and relatability, which explains its use in practices such as music therapy. However, research on music and productivity remains more divided. Highly stimulating music appears to reduce focus in many cases, while calmer instrumental music often produces results similar to silence. Because enjoyment and performance are not identical concepts, we must carefully define how productivity and effectiveness are measured when analyzing music’s effects.

#### Chi-square test of association between patients' favorite genre and mental health conditions

Based on the results of the p-values for each association in the chi-square test below, we can see that depression is the only significant finding with a p-value of 0.031, below the standard α = 0.05 threshold for statistical significance. This means that the distribution of favorite genres differs meaningfully between people who report depression and those who don't.

However, because its Cramer's V value (for effect size) is 0.19, it can be inferred that although the association between Depression and favorite genre is statistically significant, the strength of the relationship is modest in which genre explains a small portion of variance in depression presence.

Ultimately, the chi-square test on these features tell us that there is an association between Depression and favorite genre, but it does not prove that there is any causality between the features. It is also worth noting that with four simultaneous tests at α = 0.05, there is approximately a 19% chance of at least one false positive occurring by chance alone, meaning the Depression result should be interpreted with some caution.

```python
import pandas as pd
from math import sqrt
from scipy.stats import chi2_contingency

CLEANED_FILE_PATH = "data/01-interim/MxMH_Survey_cleaned.csv"
ENCODED_FILE_PATH = "data/02-processed/MxMH_Survey_encoded.csv"
encoded_df = pd.read_csv(ENCODED_FILE_PATH)

fav_df = pd.read_csv(CLEANED_FILE_PATH)
mh_cols_local = [c for c in ['Anxiety', 'Depression', 'Insomnia', 'OCD'] if c in fav_df.columns]
for c in mh_cols_local:
    fav_df[c] = pd.to_numeric(fav_df[c], errors='coerce')
    fav_df[c + '_present'] = (fav_df[c] > 0).astype(int)
mh_cols_local = [c for c in ['Anxiety', 'Depression', 'Insomnia', 'OCD'] if c in fav_df.columns]
genre_col = 'Fav genre'

chi_rows = []
for mh in mh_cols_local:
    present_col = mh + '_present'
    table = pd.crosstab(fav_df[genre_col], fav_df[present_col])
    chi2, p, dof, expected = chi2_contingency(table)
    n_obs = table.to_numpy().sum()
    cramer_v = sqrt(chi2 / (n_obs * min(table.shape[0] - 1, table.shape[1] - 1)))
    chi_rows.append({
        'condition': mh,
        'chi2': chi2,
        'p_value': p,
        'cramers_v': cramer_v,
        'dof': dof,
    })

chi_df = pd.DataFrame(chi_rows).sort_values('p_value')
print('\nChi-square association between favorite genre and mental-health presence:')
print(chi_df.to_string(index=False))
```

#### Compare General Listening to Mental Health Severity with Linear Regression Models

When comparing the hours a day someone listens to music with their perceived mental health scores, we can see a positive slope in all four graphs.

Further detail into the slope, intercept, R-squared value, and p-values of these graphs reveal that we can attribute some association for depression, insomnia, and OCD albeit weak. The p-value for Hours Listened v. Anxiety was the only one larger than the preferred 0.05 score thus its slope was statistically insignificant. Additionally, the R-squared values for all graphs was low, pointing out that the predicibility of someone's mental health scores have much to do with variables outside of music enjoyment.

```python
CLEANED_FILE_PATH = "data/01-interim/MxMH_Survey_cleaned.csv"
df = pd.read_csv(CLEANED_FILE_PATH)

#loop to make linear regression models
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(14, 10))

axes_flat = axes.flatten()

mh_cols = ['Anxiety', 'Depression', 'Insomnia', 'OCD']

# 2. Loop through your conditions and match them to the correct grid slot
for i, disorder in enumerate(mh_cols):
    if disorder in df.columns:
        # Select the specific subplot slot for this iteration
        current_ax = axes_flat[i]
        
        # Draw the regression plot, passing 'ax=current_ax'
        sns.regplot(
            data=df, 
            x='Hours per day', 
            y=disorder, 
            ax=current_ax,
            scatter_kws={'alpha': 0.4, 'color': '#1f77b4'}, 
            line_kws={'color': 'red', 'linewidth': 2}
        )

        current_ax.set_title(f'Music Listening Hours vs. {disorder} Severity', fontsize=12, fontweight='bold', pad=10)
        current_ax.set_xlabel('Hours Spent Listening to Music Per Day', fontsize=10)
        current_ax.set_ylabel(f'{disorder} Affinity Score (0-10)', fontsize=10)
        current_ax.set_ylim(-0.5, 10.5) 

        sns.despine(ax=current_ax)
        
plt.tight_layout()
plt.show()

from scipy.stats import linregress

print(f"{'Disorder':<12} | {'Slope (m)':<10} | {'Intercept (b)':<14} | {'R-squared':<10} | {'P-value':<8}")
print("-" * 65)

for disorder in mh_cols:
    if disorder in df.columns:
        clean_data = df[['Hours per day', disorder]].dropna()
        
        # X is independent (Hours), Y is dependent (Disorder Score)
        x = clean_data['Hours per day']
        y = clean_data[disorder]
        
        # Calculate the linear regression metrics
        slope, intercept, r_value, p_value, std_err = linregress(x, y)
        
        print(f"{disorder:<12} | {slope:<10.4f} | {intercept:<14.4f} | {r_value**2:<10.4f} | {p_value:<8.4f}")
```

#### BPM Distribution and Jittered Clustering Patterns Across Severity Tiers

To explore whether music tempo (BPM) relates to mental health severity, we visualize BPM distributions across six favorite genres, segmented by three severity tiers (Low: 0–2, Moderate: 3–5, Severe: 6–10) for each of the four mental health conditions.

The most notable finding is that BPM distributions remain largely consistent across severity tiers within each genre, meaning that as self-reported mental health severity increases, participants do not appear to favor faster or slower tempos. This is a meaningful null result, suggesting that tempo preference may be more of a personal preference than an association with mental health state.

Genre appears to be a stronger factor of BPM than severity. EDM listeners correlate around 120–130 BPM regardless of condition or severity tier, while Classical listeners correlate lower with more spread. Metal consistently shows the widest interquartile ranges, indicating greater BPM variance among its listeners. Rock is significant for its frequent high-BPM outliers across nearly every severity and condition combination.

An important limitation of this visualization is that many genre-severity-condition cells likely contain very few observations, meaning some box plots may reflect small samples rather than robust distributional patterns. Per-cell sample sizes were not verified, and findings from this plot should be treated as exploratory rather than conclusive.

```python
import warnings
import pandas as pd
import matplotlib.pyplot as plt

FILE_PATH = "data/01-interim/MxMH_Survey_cleaned.csv"
df = pd.read_csv(FILE_PATH)

df_clean = df.dropna(subset=['BPM', 'Fav genre'])
# Strip out joke entries to preserve scale uniformity across the grid
df_clean = df_clean[(df_clean['BPM'] >= 20) & (df_clean['BPM'] <= 300)]

# Standardize the Y-axis categories by focusing on the top overall genres
top_genres = df_clean['Fav genre'].value_counts().index[:6]
df_matrix = df_clean[df_clean['Fav genre'].isin(top_genres)]

# 2. Define our matrix dimensions
conditions = ['Anxiety', 'Depression', 'Insomnia', 'OCD']
severity_labels = ['Severe (6-10)', 'Moderate (3-5)', 'Low (0-2)']

nrows = len(severity_labels) # 3
ncols = len(conditions)      # 4

# sharey=True forces every single graph to use the exact same genre listing on the Y-axis
# sharex=True keeps the 20-300 BPM scale completely aligned for instant visual comparison
fig, axes = plt.subplots(nrows, ncols, figsize=(24, 16), sharex=True, sharey=True)

# 3. Populate the Matrix Grid
for r, sev in enumerate(severity_labels):
    for c, cond in enumerate(conditions):
        ax = axes[r, c]
        
        # Filter the data for this exact cell based on severity tier definitions
        if r == 0:     # Severe
            cell_df = df_matrix[df_matrix[cond] >= 6]
        elif r == 1:   # Moderate / Average
            cell_df = df_matrix[(df_matrix[cond] >= 3) & (df_matrix[cond] <= 5)]
        else:          # Low
            cell_df = df_matrix[df_matrix[cond] <= 2]
            
        # If a cell has data, plot the distribution
        if not cell_df.empty:
            # Boxplot for the core distribution structure
            sns.boxplot(
                data=cell_df,
                y='Fav genre',
                x='BPM',
                ax=ax,
                order=top_genres,
                palette='Set3',
                width=0.4,
                fliersize=0
            )
            # Stripplot overlay to reveal raw cluster density
            sns.stripplot(
                data=cell_df,
                y='Fav genre',
                x='BPM',
                ax=ax,
                order=top_genres,
                color='#2c3e50',
                size=4,
                alpha=0.4,
                jitter=0.2
            )
            
        # --- Metadata and Layout Cleanup ---
        # Set column titles only on the very top row
        if r == 0:
            ax.set_title(cond, fontsize=16, fontweight='bold', pad=12)
            
        # Set row headers only on the leftmost column
        if c == 0:
            ax.set_ylabel(sev, fontsize=14, fontweight='bold', color='#2c3e50')
        else:
            ax.set_ylabel('')
            
        # Label the X-axis only on the bottom row to prevent text crowding
        if r == 2:
            ax.set_xlabel('Tempo (BPM)', fontsize=12)
        else:
            ax.set_xlabel('')
            
        ax.grid(axis='x', linestyle='--', alpha=0.3)

# 4. Super-title and export
plt.suptitle('BPM Distribution & Jittered Clustering Patterns Across Severity Tiers', 
             fontsize=20, fontweight='bold', y=0.98)
plt.tight_layout()
plt.savefig('mental_health_bpm_matrix.png', dpi=300, bbox_inches='tight')
```

#### T-Tests for Reaction time, Mood, Arousal, and Thought probe

These are four t-tests done on each variable: Mood, Arousal, Thought Probe, and Reaction. We see that for our p-tests the results are meaningful for **Mood, Arousal,** and **Thought probe**. The p-tests results for Reaction time are very large, indicating it is not statistically significant.

For t-test we see that it is extremely low for reaction time as well compared to other variables. Despite low variance in mean, we can conclude that for mood, arousal, and thought probe they were all affected by music trials while reaction time was not.


```python
from scipy.stats import ttest_rel
import pandas as pd

FILE_PATH = "data/01-interim/KL_Dataset_cleaned.csv"
df = pd.read_csv(FILE_PATH)


variables = ['Reaction time', 'Mood', 'Arousal', 'Thought probe ']

for var in variables:
    # Optional: use cleaned reaction time only for Reaction time
    if var == 'Reaction time':
        temp_df = df[df['Reaction time'] < 10].copy()
    else:
        temp_df = df.copy()

    # Average each variable by participant and music condition
    summary = (
        temp_df
        .dropna(subset=[var])
        .groupby(['ID', 'Music-present'])[var]
        .mean()
        .reset_index()
    )

    # Make one column for No Music and one for Music
    wide = summary.pivot(
        index='ID',
        columns='Music-present',
        values=var
    )

    wide.columns = ['No_Music', 'Music']
    wide = wide.dropna()

    # Paired t-test
    result = ttest_rel(
        wide['Music'],
        wide['No_Music']
    )

    print(var)
    print(f"No Music mean: {wide['No_Music'].mean():.3f}")
    print(f"Music mean: {wide['Music'].mean():.3f}")
    print(f"t({result.df}) = {result.statistic:.3f}, p = {result.pvalue:.3f}")
    print("\n")
```
